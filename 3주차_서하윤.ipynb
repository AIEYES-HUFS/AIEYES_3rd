```python
Chapter 7 — Dimensionality Reduction
1	장점: 학습 속도·메모리 감소, noise/redundancy 제거, 시각화 / 단점: 정보 손실
2	차원이 증가할수록 공간이 급격히 커지고 데이터가 희소해지는 차원의 저주
3	완벽한 원상복구는 불가능. 일부 정보가 손실되며 PCA 등으로 근사 복원 가능
4	가능하지만 고도로 비선형적인 데이터에는 PCA가 적합하지 않을 수 있음
5	데이터의 분산 구조에 따라 달라지므로 정확한 차원 수는 미리 알 수 없음
6	Regular PCA: 일반적인 경우 / Incremental PCA: 매우 큰 데이터 / Randomized PCA: 빠른 PCA / Random Projection: 고차원 데이터를 빠르게 축소
7	Reconstruction error, 차원 축소 후의 ML 성능, 계산 시간 등을 비교
8	가능함. 두 차원 축소 알고리즘을 연속으로 사용할 수 있지만 항상 효과적인 것은 아님
9	MNIST → Random Forest → PCA(95%) → Random Forest 재학습 → 학습시간·정확도 비교, 이후 SGDClassifier도 비교
10	MNIST 5,000개 → t-SNE 2차원 변환 → 숫자별 cluster 시각화, PCA·LLE·MDS와 비교
```
