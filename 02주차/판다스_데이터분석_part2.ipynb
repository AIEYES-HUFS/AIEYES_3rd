**2-1 HTML 웹 페이지에서 표 속성 가져오기**

- 판다스 read_html() 함수는 HTML 웹 페이지에 있는 table 태그에서 표 형식의 데이터를 모두 찾아서 데이터프레임으로 변환함
- 표 데이터들은 각각 별도의 데이터프레임으로 변환되기 때문에 여러 개의 데이터프레임을 원소로 갖는 리스트가 반환됨

```python
# 예제 2-4: 웹에서 표 정보 읽기

import pandas as pd

# HTML 파일 경로 or 웹 페이지 주소를 url 변수에 저장
url = './sample.html'

#HTML 웹페이지의 표를 가져와서 데이터프레임으로 변환
tables = pd.read_html(url)

# 표의 개수 확인
print(len(tables))
print("\n")

# tablesm 리스트의 원소를 iteration하면서 각각 화면 출력
for i in range(len(tables)):
  print("tables[%s]" % i)
  print(tables[i])
  print("\n")

# 파이썬 패키지 정보가 들어있는 두번째 데이터프레임을 선택하여 df 변수에 저장
df = tables[1]

# 'name' 열을 인덱스로 지정
df.set_index(['name'], inplace = True)
print(df)

```

**웹 스크래핑**

- 스크래핑한 내용을 파이썬 리스트, 딕셔너리 등으로 정리한 뒤 DataFrame() 함수에 리스트나 딕셔너리 형태로 전달해 데이터프레임으로 변환
- 왼쪽의 딕셔너리 키는 열 이름이 되고, 오른쪽 리스트는 열 데이터가 됨

```python
# 예제 2-5: 미국 ETF 리스트 가져오기 (최신 위키피디아 대응)

import requests    # 정규표현식 검색
import pandas as pd    # 데이터프레임 생성
import re     # HTML 분석

url = "https://en.wikipedia.org/wiki/List_of_American_exchange-traded_funds"

headers = {
    "User-Agent": "Mozilla/5.0"   # 파이썬 봇 차단 방지 -> 브라우저처럼 접ㅈ속
}

resp = requests.get(url, headers=headers)
resp.raise_for_status()

text = resp.text

# ETF 정보 추출
pattern = r'([A-Za-z0-9&.,\'\-\s]+)\s*\(NYSE\s*Arca[:\s]+([A-Z]+)\)'

# ([A-Za-z0-9&.,\'\-\s]+) : etf 이름 추출
# ([A-Z]+) : 티커 추출

matches = re.findall(pattern, text)

etfs = {}

for name, ticker in matches:

    name = re.sub(r'\s+', ' ', name).strip()

    etfs[ticker] = [
        "NYSE Arca",
        name
    ]

print(etfs)
print("\n")

df = pd.DataFrame(etfs)

print(df)
```

**여러 개의 데이터프레임을 하나의 excel 파일로 저장**

- ExcelWriter() 함수는 Excel 워크북 객체를 생성
- 데이터프레임에 to_excel() 메소드를 적용할 때 삽입하려는 워크북 객체를 인자로 전달
- 데이터프레임을 삽입하는 시트 이름을 다르게 설정하면, 같은 excel 파일의 서로 다른 시트에 여러 데이터프레임을 구분하여 저장

```python
# 예제 2-10: excelwriter() 활용

import pandas as pd

# 판다스 DataFrame() 함수로 데이터프레임 변환. 변수 df1, df2에 저장
data1 = {'name': ['Jerry', 'Riah', 'Pual'],
        'algol': ['A', 'A+', 'B'],
        'basic' : ["c", "B", "B+"],
        'C++': ["B+", "C", "C+"]}

data2 = {'c0': [1,2,3],
         'c1':[4,5,6],
         'c2':[7,8,9],
         'c3':[10,11,12],
         'c4':[13,14,15]}

df1 = pd.DataFrame(data1)
df1.set_index('name', inplace= True)
print(df1)
print('\n')

df2 = pd.DataFrame(data2)
df2.set_index('c0', inplace=True)
print(df2)

# df1을 'sheet1'으로, df2를 'sheet2'로 저장(Excel 파일명은 "df_excelwriter.xlsx")
writer = pd.ExcelWriter("./df_excelwriter.xlsx")
df1.to_excel(writer, sheet_name = 'sheet1')
df2.to_excel(writer, sheet_name = 'sheet2')
writer.close()
```
