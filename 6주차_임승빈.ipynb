# Chapter 11. 심층 신경망 훈련


## 1. 신경망이 학습한다는 것은 무엇일까?

신경망은 입력된 정보를 여러 단계에 걸쳐 계산하고 답을 내놓는 모델이다. 예를 들어 옷 사진을 입력하면 티셔츠인지 바지인지 예측한다. 신경망 안에서 계산을 담당하는 작은 단위를 뉴런이라고 하고, 뉴런이 모인 계산 단계를 층이라고 한다. 입력과 출력 사이에 있는 계산 단계는 은닉층이라고 부른다. 은닉층이 여러 개 있는 신경망을 심층 신경망이라고 한다.

각 연결에는 정보의 영향력을 조절하는 값인 **가중치**가 있다. 어떤 정보에 더 큰 영향을 줄 것인지 정하는 숫자라고 생각하면 된다. **편향**은 계산 결과의 기준을 조금 높이거나 낮추는 값이다. 가중치와 편향처럼 학습하면서 바뀌는 값을 파라미터라고 한다.

모델은 먼저 답을 예측하고 실제 정답과 비교한다. 예측이 얼마나 틀렸는지를 숫자로 나타낸 것이 **손실**이다. 학습은 이 손실이 작아지도록 가중치와 편향을 조금씩 바꾸는 과정이다.

이때 각 값을 어느 방향으로 바꾸면 좋을지 알려 주는 것이 **기울기**다. 산에서 경사를 보고 내려갈 방향을 찾는 것과 비슷하다. 출력에서부터 앞쪽 층으로 거슬러 올라가며 기울기를 구하는 과정을 **역전파**라고 한다.


## 2. 기울기가 너무 작거나 커지면 생기는 문제

신경망이 깊어지면 역전파 과정에서 여러 층의 값이 반복해서 곱해진다. 이 때문에 기울기가 지나치게 작아지거나 커질 수 있다.

**기울기 소실**은 기울기가 너무 작아지는 문제다. 0.5를 여러 번 곱하면 점점 0에 가까워지는 것과 비슷하다. 앞쪽 층에 전달되는 기울기가 너무 작으면 가중치가 거의 바뀌지 않는다. 틀린 답을 보고도 앞쪽 층이 충분히 배우지 못하는 셈이다.

**기울기 폭주**는 반대로 기울기가 너무 커지는 문제다. 2를 여러 번 곱하면 값이 빠르게 커진다. 기울기가 너무 크면 가중치를 한 번에 지나치게 많이 바꾸게 되어 학습이 불안정해질 수 있다.

따라서 깊은 신경망을 만들 때는 층의 개수만 볼 것이 아니라, 기울기가 각 층에 적절한 크기로 전달되는지도 생각해야 한다. 초기화, 활성화 함수, 정규화 층, 기울기 클리핑은 이 문제를 줄이기 위한 방법들이다.


## 3. 처음 가중치를 정하는 방법

**초기화**는 학습을 시작하기 전에 가중치와 편향의 첫 값을 정하는 것이다. 처음부터 가중치가 너무 크거나 작으면 층을 지나는 정보의 크기도 불안정해질 수 있다. 그래서 입력과 출력의 개수, 사용하는 활성화 함수에 맞춰 적당한 범위에서 가중치를 무작위로 정한다.

모든 뉴런의 가중치를 똑같이 시작하는 것은 좋지 않다. 같은 입력에 대해 같은 계산을 하고 같은 방식으로 바뀌면, 여러 뉴런이 있어도 서로 다른 특징을 배우기 어렵기 때문이다. 여러 사람이 모둠 활동을 하는데 모두 똑같은 일만 하는 상황과 비슷하다. 가중치를 다르게 시작하면 서로 다른 특징을 배울 수 있는 출발점을 만든다.

대표적인 초기화 방법에는 **Glorot 초기화**, **He 초기화**, **LeCun 초기화**가 있다. Glorot은 tanh 같은 함수와, He는 ReLU 계열과 연결해서 이해하면 좋다. LeCun은 뒤에서 나오는 SELU의 조건과 함께 기억하면 된다. 이 방법들은 연결 수 등을 고려하여 처음 가중치가 퍼지는 정도를 조절한다.

가중치와 달리 편향은 0으로 시작해도 괜찮다. 가중치가 이미 서로 다르게 정해져 있다면 편향이 같다는 이유만으로 모든 뉴런이 똑같아지지는 않는다.

초기화는 좋은 출발점을 만드는 방법이다. 처음 값을 잘 정했다고 해서 학습이 끝날 때까지 모든 문제가 자동으로 해결되는 것은 아니다.


## 4. 활성화 함수는 어떤 역할을 할까?

**활성화 함수**는 뉴런이 계산한 값을 다음 층으로 보내기 전에 바꾸는 함수다. 신경망이 단순한 직선 관계를 넘어 복잡한 규칙을 배우도록 해 준다. 이런 변화 없이 선형 계산만 여러 번 쌓으면, 결국 하나의 선형 계산으로 합칠 수 있어 표현력이 제한된다.

**ReLU**는 양수는 그대로 보내고 음수는 0으로 바꾼다. 규칙이 단순해서 계산이 빠르다. 하지만 어떤 뉴런이 계속 음수만 받으면 기울기도 계속 0이 되어 학습에 거의 참여하지 못할 수 있다. 이를 죽은 ReLU 문제라고 한다.

**Leaky ReLU**는 음수를 완전히 막지 않고 작은 비율로 통과시킨다. 문을 완전히 닫지 않고 조금 열어 두는 것처럼, 음수 구간에서도 기울기가 흐를 길을 남겨 둔다.

**ELU**도 음수를 전부 0으로 만들지는 않는다. 음수 구간을 곡선으로 처리해서 음수 출력이 나올 수 있도록 한다. 다만 단순한 ReLU보다 계산이 복잡하다.

**SELU**는 특정 조건에서 층을 지나도 값들의 평균과 퍼진 정도가 일정한 수준에 가까워지도록 만든 함수다. 이 성질을 자기 정규화라고 한다. 입력의 크기를 미리 맞추고, LeCun 초기화를 사용하며, 적절한 층 구조를 쓰는 등의 조건이 필요하다. SELU 하나만 넣으면 모든 신경망이 저절로 안정되는 것은 아니다.


## 5. 더 부드러운 활성화 함수와 SwiGLU

Chapter 11에서는 GELU, Swish, Mish처럼 값을 부드럽게 바꾸는 함수도 다룬다. ReLU가 음수인지 양수인지에 따라 처리한다면, 이 함수들은 입력의 크기에 따라 전달되는 양을 더 부드럽게 조절한다.

**GELU**는 입력값에 따라 그 값을 얼마나 통과시킬지 부드럽게 조절한다. **Swish**는 입력값에 Sigmoid 함수로 구한 비율을 곱한다. Sigmoid는 값을 0과 1 사이로 바꾸는 함수다. Swish의 대표적인 설정은 PyTorch에서 SiLU라는 이름으로 제공된다. **Mish** 역시 부드러운 곡선을 사용하며 음수도 일부 전달한다.

**ReLU²**는 ReLU의 결과를 제곱한 것이다. 음수는 0이 되고, 양수는 제곱된다. 이름이 복잡해 보여도 어떤 계산을 하는지 나누어 보면 이해하기 쉽다.

**SwiGLU**는 정보를 두 갈래로 나누고, 한쪽에 Swish를 적용한 뒤 다른 쪽과 같은 위치끼리 곱한다. 한 갈래가 다른 갈래의 전달량을 조절하는 역할을 하는 셈이다. 여기서 gate는 정보가 얼마나 전달될지 조절하는 장치라는 뜻으로 이해했다.

활성화 함수를 고를 때는 복잡한 함수가 항상 더 좋다고 생각하면 안 된다. 기울기가 잘 전달되는지, 계산이 얼마나 필요한지, 실제 검증 성능이 좋아지는지를 함께 살펴봐야 한다.


## 6. 값의 크기를 맞추는 정규화 층

학습 중에는 가중치가 계속 바뀌기 때문에 각 층에서 나오는 값의 크기도 달라진다. 값이 지나치게 커지거나 작아지면 다음 층의 학습이 어려워질 수 있다. **정규화 층**은 중간 계산값의 크기를 조절하여 학습을 돕는다.

정규화에서는 평균과 값들이 퍼진 정도를 이용해 크기를 맞춘다. 그 뒤에는 모델이 필요한 크기와 위치를 다시 학습할 수 있도록 조절용 파라미터도 둔다. 무조건 같은 숫자로 만드는 것이 아니라, 값들의 상대적인 차이를 유지하면서 크기를 다루기 쉽게 만드는 것이다.

**Batch Normalization**, 줄여서 BatchNorm은 여러 샘플을 한 묶음으로 보고 통계를 구한다. 이 묶음을 배치라고 한다. 완전연결 층에서는 배치 안의 같은 특성끼리 비교하여 평균과 퍼진 정도를 계산한다. 여러 학생의 같은 과목 점수를 모아 기준을 잡는 상황과 비슷하다.

BatchNorm은 학습할 때 현재 배치의 통계를 사용한다. 일반적인 설정에서는 평가할 때 학습 중 쌓아 둔 통계를 사용한다. 따라서 학습 모드와 평가 모드를 구분해야 한다. 배치가 너무 작으면 기준을 잡을 자료가 적어서 통계가 불안정할 수 있다.

**Layer Normalization**, 줄여서 LayerNorm은 다른 샘플을 함께 보지 않고 한 샘플 안의 지정된 특성들을 기준으로 계산한다. 앞의 비유를 이어 가면 한 학생 안에서 정해진 여러 과목의 점수를 보고 기준을 잡는 것과 비슷하다. 실제로는 개발자가 지정한 특성 차원에 적용한다.

LayerNorm은 다른 샘플의 수에 덜 의존하며, 학습과 평가 모두 현재 입력의 통계를 사용한다. 두 방법의 핵심 차이는 이름보다 **어떤 값들을 한데 모아 기준을 계산하는가**에 있다.


## 7. 너무 큰 기울기를 제한하는 방법

**기울기 클리핑**은 기울기가 정해 둔 한도보다 커지면 줄이는 방법이다. 내리막길에서 속도가 너무 빨라질 때 브레이크를 거는 것과 비슷하다. 가중치를 한 번에 지나치게 많이 바꾸는 일을 줄여 준다.

대표적인 방법은 전체 기울기의 크기를 측정한 뒤, 한도를 넘으면 같은 비율로 줄이는 것이다. 이 경우 이동하려는 방향은 유지하면서 이동량을 줄일 수 있다.

적용 순서도 중요하다. 먼저 역전파로 기울기를 계산하고, 그 기울기를 제한한 다음, 가중치를 바꿔야 한다. PyTorch에서는 `loss.backward()` 뒤, `optimizer.step()` 앞에 클리핑을 넣는다.

이 방법은 너무 큰 기울기를 다루는 데 도움이 된다. 이미 너무 작아진 기울기를 다시 키워 주는 방법은 아니므로 기울기 소실과 폭주의 해결책을 구분해야 한다.


## 8. 이미 배운 내용을 활용하는 전이학습

**전이학습**은 다른 문제를 풀며 배운 내용을 새 문제에 활용하는 방법이다. 자전거를 타며 익힌 균형 감각이 다른 운동을 배울 때 도움이 되는 것과 비슷하다. 새 과제의 데이터가 적을 때 특히 유용할 수 있지만, 두 과제가 얼마나 관련 있는지에 따라 효과는 달라진다.

원본 예제에서는 옷 사진을 분류하는 Fashion MNIST를 사용한다. 먼저 티셔츠와 풀오버를 제외한 여덟 종류의 옷을 구분하도록 학습한다. 그다음에는 티셔츠와 풀오버를 구분하는 새 문제를 만든다. 새 문제의 훈련 사진은 두 종류를 합쳐 20장만 사용한다.

기존 모델에서 마지막 정답 결정 층은 바꾸고, 앞에서 특징을 계산하던 층들은 재사용한다. 처음에는 재사용하는 부분의 가중치를 고정하고 새로 붙인 출력층부터 학습한다. 이렇게 기존 가중치가 바뀌지 않도록 하는 것을 **동결**이라고 한다.

새 출력층이 어느 정도 적응하면 기존 층 일부도 다시 학습하게 할 수 있다. 이를 **미세 조정**이라고 한다. 이미 배운 내용이 급격히 바뀌지 않도록 보통 작은 학습률로 조금씩 조정한다.

주의할 점은 가중치를 고정하는 것과 평가 모드로 바꾸는 것이 다르다는 것이다. 가중치를 동결해도 BatchNorm의 통계나 Dropout의 동작은 별도로 관리해야 할 수 있다.


## 9. 가중치를 더 효율적으로 바꾸는 옵티마이저

**옵티마이저**는 계산된 기울기를 보고 가중치를 어떻게 바꿀지 정하는 방법이다. **학습률**은 한 번 바꿀 때 얼마나 크게 움직일지 정하는 값이다. 산을 내려갈 때 옵티마이저가 이동 방법이라면, 학습률은 걸음의 크기라고 생각할 수 있다.

기본적인 경사 하강법은 현재 위치에서 구한 기울기를 보고 손실이 줄어드는 방향으로 움직인다. **SGD**는 훈련 데이터의 일부를 이용해 이런 갱신을 반복하는 방식이다. 실제 학습에서는 보통 작은 데이터 묶음인 미니배치를 사용한다.

**Momentum**은 지금의 기울기뿐 아니라 이전에 움직이던 방향도 반영한다. 공이 굴러가던 방향으로 계속 움직이려는 관성과 비슷하다. 일관된 방향으로 더 빠르게 이동하고 흔들림을 줄이는 데 도움이 된다. 하지만 관성을 너무 크게 주면 좋은 지점을 지나쳐 여러 번 오갈 수 있다.

**Nesterov**는 관성으로 조금 더 이동했을 위치를 고려하여 방향을 보정하는 방식이다. Momentum을 좀 더 앞을 내다보도록 바꾼 방법으로 이해했다.

**AdaGrad**는 가중치마다 지금까지의 기울기를 모아서 이동 크기를 다르게 조절한다. 많이 변했던 방향에서는 걸음을 줄이는 식이다. 하지만 기록이 계속 쌓이면서 학습 후반에 걸음이 너무 작아질 수 있다.

**RMSProp**은 오래된 기록의 영향은 점차 줄이고 최근 기울기를 더 반영한다. AdaGrad처럼 과거 기록이 계속 누적되어 걸음이 지나치게 작아지는 문제를 완화하려는 방법이다.


## 10. Adam과 그 변형

**Adam**은 기울기가 향하는 방향의 흐름과 기울기 크기의 흐름을 함께 활용한다. Momentum처럼 이전 방향을 참고하면서, RMSProp처럼 가중치별로 걸음의 크기도 조절한다고 이해하면 된다.

**Adamax**는 Adam에서 기울기의 크기를 다루는 방식을 바꾼 변형이다. **NAdam**은 Adam에 Nesterov의 아이디어를 더한 방법이다. 이 단계에서는 세부 계산을 전부 외우기보다 어떤 방법을 바탕으로 만들어졌는지 연결해서 기억했다.

**AdamW**는 Adam의 학습 과정과 가중치를 조금씩 줄이는 처리를 분리한 방법이다. 가중치를 줄이는 처리는 모델이 훈련 자료에 지나치게 맞춰지는 것을 줄이는 데 쓰인다. 뒤에서 정리하는 weight decay와 연결된다.

옵티마이저가 자동으로 걸음 크기를 조절하더라도 처음 정하는 학습률은 여전히 중요하다. Adam을 쓴다는 이유만으로 학습률을 아무렇게나 정해도 되는 것은 아니다. 같은 검증 데이터를 기준으로 여러 설정을 비교해야 한다.


## 11. 학습률을 학습 도중 바꾸기

학습률이 너무 크면 좋은 지점을 계속 지나칠 수 있고, 너무 작으면 배우는 데 오래 걸린다. 그래서 처음부터 끝까지 같은 학습률을 쓰는 대신 학습 과정에 맞춰 바꾸기도 한다. 이를 **학습률 스케줄링**이라고 한다.

**지수 감소**는 일정 단계마다 학습률에 같은 비율을 곱해 줄이는 방법이다. 예를 들어 매번 이전 값의 90%로 바꾼다. 학습이 진행될수록 걸음이 점점 작아진다.

**성능 기반 감소**는 검증 성능이 한동안 좋아지지 않을 때 학습률을 낮춘다. 더 작은 걸음으로 주변을 살펴보게 하는 것이다. 손실을 기준으로 삼으면 작아지는 것이 개선이고, 정확도를 기준으로 삼으면 커지는 것이 개선이므로 설정을 맞춰야 한다.

**Warmup**은 작은 학습률로 시작해서 점차 높이는 준비 단계다. 학습 초기에 가중치가 급격히 바뀌는 것을 줄이는 데 도움이 될 수 있다.

**코사인 감소**는 학습률이 매끄러운 곡선을 따라 낮아지도록 하는 방법이다. **코사인 재시작**은 학습률을 낮춘 뒤 다시 높이고 이 과정을 반복한다. 다시 시작한다는 말은 학습률에 대한 것으로, 배운 가중치를 전부 지운다는 뜻은 아니다.

**1cycle**은 학습률을 올렸다가 다시 낮추는 흐름을 한 번의 학습 과정에 넣는다. 학습 후반에는 더 작은 값으로 세밀하게 조정한다.

전체 훈련 데이터를 한 번 다 보는 것을 **epoch**라고 한다. 스케줄러에 따라 epoch마다 학습률을 바꾸기도 하고 배치마다 바꾸기도 한다. 특히 OneCycleLR은 보통 각 배치에서 가중치를 갱신한 다음 호출한다. 어떤 주기로 바꾸는지도 방법의 일부라는 점을 기억했다.


## 12. 과적합과 L1·L2 규제

**과적합**은 모델이 훈련 자료에는 잘 맞지만 새로운 자료에는 잘 맞지 않는 상태다. 연습문제의 답만 외워서 비슷하지만 새로운 문제를 풀지 못하는 상황과 비슷하다.

**규제**는 이런 문제를 줄이기 위해 학습에 제한이나 추가 조건을 주는 방법이다. 모델이 훈련 자료의 작은 우연까지 지나치게 따라가지 않도록 돕는다. 앞에서 본 정규화 층은 값의 크기를 조절하는 층이고, 여기서 말하는 규제는 과적합을 줄이기 위한 방법이라는 점을 구분했다.

**L1 규제**는 가중치의 절댓값을 모아서 손실에 추가한다. 예측을 잘하는 것뿐 아니라 가중치를 작게 유지하는 것도 학습 목표에 넣는 것이다. 일부 가중치가 0에 가까워지는 방향으로 유도하므로, 적은 연결에 집중하는 모델을 만드는 데 도움이 된다. 다만 실제 계산에서 작은 가중치가 모두 정확히 0이 되는 것은 아니다.

**L2 규제**는 가중치를 제곱한 값을 모아서 손실에 추가한다. 큰 가중치일수록 부담이 더 커지므로 전체적으로 가중치가 너무 커지는 것을 막는 데 도움이 된다. L1이 일부 연결을 줄이는 성질과 연결된다면, L2는 가중치 크기를 전반적으로 줄이는 성질에 초점을 맞춰 이해했다.

**Weight decay**는 가중치를 갱신할 때 조금씩 줄이는 방법이다. 단순한 SGD에서는 설정을 맞추면 L2와 같은 형태의 갱신이 될 수 있다. 하지만 Adam처럼 기울기 기록을 이용하는 방법에서는 손실에 L2를 넣는 것과 가중치를 따로 줄이는 것이 같지 않다. AdamW는 이 축소를 따로 처리한다.

규제를 너무 강하게 주면 필요한 특징까지 충분히 배우지 못할 수 있다. 규제 역시 무조건 많이 적용하는 것보다 검증 성능을 보며 적절한 정도를 찾는 것이 중요하다.


## 13. Dropout으로 특정 뉴런에 대한 의존 줄이기

**Dropout**은 학습할 때 일부 뉴런의 출력을 무작위로 0으로 만드는 방법이다. 매번 일부 정보가 빠진 상태로 배우게 하여 특정 뉴런에만 지나치게 의존하지 않도록 한다.

모둠 과제에서 항상 한 사람에게만 의존하지 않도록, 연습할 때마다 일부 구성원 없이도 과제를 해 보는 상황과 비슷하다. 특정 정보가 없어도 다른 정보로 답을 구하는 능력을 기르게 하는 것이다.

Dropout에서 `p=0.2`라면 출력이 0으로 제거될 확률이 20%라는 뜻이다. 20%를 남긴다는 뜻이 아니다. PyTorch의 일반 Dropout은 학습 중 남아 있는 값의 크기를 보정하여 평균적인 크기가 갑자기 줄지 않게 한다.

일반적으로 실제 예측을 할 때는 Dropout을 끈다. 학습 모드에서는 무작위 제거가 작동하고 평가 모드에서는 제거하지 않는다. SELU의 자기 정규화 성질을 이용하는 모델에서는 일반 Dropout 대신 **AlphaDropout**을 사용하여 값들의 평균과 퍼진 정도가 유지되도록 돕는다.


## 14. 여러 번 예측하는 MC Dropout

**MC Dropout**은 예측할 때도 Dropout을 켜고 같은 입력에 대해 여러 번 답을 구하는 방법이다. 매번 빠지는 출력이 다르기 때문에 조금씩 다른 예측을 얻게 된다.

예를 들어 같은 옷 사진을 여러 번 예측한 뒤 각 종류일 확률을 평균할 수 있다. 매번 비슷하게 예측하면 결과가 비교적 일관된 것이고, 예측이 크게 달라지면 불확실성을 의심해 볼 수 있다. 다만 그 차이가 실제로 틀릴 확률을 정확히 알려 주는 것은 아니다.

이 방법은 예측을 여러 번 해야 하므로 한 번만 예측할 때보다 계산량이 많다. 일반 Dropout과 MC Dropout의 큰 차이는 추론할 때도 무작위 제거를 사용하는지와 예측을 반복하는지에 있다.

구현할 때는 모델 전체를 학습 모드로 바꾸기보다 Dropout 층만 따로 켜는 것이 중요하다. 전체를 학습 모드로 바꾸면 BatchNorm까지 동작 방식이 달라질 수 있기 때문이다.


## 15. Max-Norm과 조기 종료

**Max-Norm**은 한 뉴런에 연결된 가중치 묶음의 전체 크기가 정해진 한도를 넘지 않도록 제한하는 방법이다. 가중치를 갱신한 뒤 너무 커진 묶음만 비율을 맞춰 줄인다.

기울기 클리핑과 비슷해 보이지만 대상이 다르다. 기울기 클리핑은 가중치를 바꾸기 전에 **기울기**를 제한한다. Max-Norm은 가중치를 바꾼 뒤 **가중치 자체**를 제한한다. 바꾸려는 양을 줄이는지, 바꾼 결과의 크기를 제한하는지로 구분하면 이해하기 쉽다.

**조기 종료**는 검증 성능이 일정 기간 좋아지지 않으면 학습을 멈추는 방법이다. 오래 공부한다고 항상 새 문제를 더 잘 푸는 것은 아니듯, 오래 학습한다고 성능이 계속 좋아지는 것은 아니다.

학습을 멈춘 마지막 순간의 모델이 가장 좋은 모델은 아닐 수 있다. 그래서 검증 성능이 가장 좋았을 때의 가중치를 저장해 두고 나중에 복원한다.

여기서 **훈련 데이터**는 직접 배우는 자료, **검증 데이터**는 학습 방법을 고르고 중간 실력을 확인하는 자료, **테스트 데이터**는 마지막에 최종 실력을 확인하는 자료다. 테스트 결과를 계속 보면서 설정을 바꾸면 최종 시험 문제를 미리 보며 연습하는 것과 비슷해질 수 있다.


## 16. PyTorch에서 헷갈리지 말아야 할 동작

`model.train()`은 모델을 학습 모드로 바꾼다. 이 명령만 실행한다고 학습이 시작되는 것은 아니다. Dropout이나 BatchNorm처럼 학습과 평가에서 행동이 달라지는 층에 현재 모드를 알려 주는 것이다.

`model.eval()`은 평가 모드로 바꾼다. 일반 Dropout을 끄고 BatchNorm이 평가 방식으로 동작하도록 한다. 하지만 기울기를 계산하는 기능까지 꺼 주지는 않는다.

`torch.no_grad()`는 그 안의 계산에 대해 기울기 추적을 하지 않도록 한다. 평가할 때 필요 없는 기록을 줄일 수 있다. 그래서 일반적인 평가에서는 `model.eval()`과 `torch.no_grad()`를 함께 사용한다.

`requires_grad=False`는 특정 파라미터의 기울기를 구하지 않도록 하는 설정이다. 전이학습에서 가중치를 동결할 때 사용한다. 평가 모드와는 역할이 다르다.

또한 분류 모델이 처음 내놓는 점수는 **로짓**이라고 한다. 아직 확률로 바꾸기 전의 값이다. PyTorch의 `CrossEntropyLoss`와 `BCEWithLogitsLoss`는 이 점수를 직접 받아 처리하므로, 그 앞에서 다시 확률로 바꾸지 않도록 주의해야 한다.
