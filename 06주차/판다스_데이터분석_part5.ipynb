## 5.1. 누락 데이터 처리

- 데이터프레임의 원소데이터값이 종종 누락됨: 데이터를 파일로 입력할 때 빠트리거나 파일 형식을 변환하는 과정에서 데이터가 소실되는 것이 주요 원인
- 일반적으로 유효한 데이터값이 존재하지 않는 누락 데이터를 NaN으로 표시
- 누락 데이터가 많아지면 데이터의 품질이 떨어지고 머신러닝 분석 알고리즘을 왜곡하는 현상 발생
- 누락 데이터의 개수를 확인하려면 dropna = False 옵션을 사용
- isnull(): 누락 데이터면 True 반환, 유효한 데이터가 존재하면 False를 반환
- notnull(): 유효한 데이터가 존재하면 True를 반환, 누락데이터면 False 반환

```python
# 예제 5-1: 누락 데이터 확인

import seaborn as sns

df = sns.load_dataset('titanic')

# deck 열의 NaN 개수 계산하기
nan_deck = df['deck'].value_counts(dropna=False)
print(nan_deck)

print("\n")
print("--------------------------------------------------------")
print("\n")

# isnull() 메소드로 누락데이터 찾기
print(df.head().isnull())

print("\n")
print("--------------------------------------------------------")
print("\n")


# notnull() 메소드로 누락데이터 찾기
print(df.head().notnull())

print("\n")
print("--------------------------------------------------------")
print("\n")

# isnull() 메소드로 누락 데이터 개수 구하기
print(df.head().isnull().sum(axis=0))
```

## 2. 누락 데이터 제거
- 열을 삭제하면 분석대상이 갖는 특성을 제거, 행을 삭제하면 분석 대상의 관측값을 제거

```python
from pandas.core.dtypes import missing
# 예제 5-2: 누락 데이터 제거

import seaborn as sns

df = sns.load_dataset('titanic')

# for 반복문으로 각 열의 NaN 개수 계산하기
missing_df = df.isnull()
for col in missing_df.columns:
    missing_count = missing_df[col].value_counts()    # 각 열의 NaN 개수 파악

    try:
      print(col, ': ', missing_count[True])    # NaN 값이 있으면 개수 출력
    except:
      print(col, ': ', 0)    # NaN 값이 없으면 0개 출력

print("\n")
print("--------------------------------------------------------")
print("\n")

# NaN값이 500개 이상인 열을 모두 삭제 - deck 열(891개 중 688개의 NaN값)
df_thresh = df.dropna(axis=1, thresh=500)
print(df_thresh.columns)

print("\n")
print("--------------------------------------------------------")
print("\n")

# age 열에 나이 데이터가 없는 모든 행 삭제 - age 열(891개 중 177개의 NaN값)
df_age = df.dropna(subset = ['age'], how = 'any', axis = 0)
print(len(df_age))
```

## 3. 평균으로 누락 데이터 바꾸기

- 누락데이터 대체값: 평균값, 최빈값 등

```python
# 예제 5-3: 평균으로 누락데이터 바꾸기

import seaborn as sns

df = sns.load_dataset('titanic')

print(df['age'].head(10))
print("\n")

# age열의 NaN 값을 다른 나이 데이터의 평균으로 변경하기
mean_age = df['age'].mean(axis=0)    # age 열의 평균 계산(NaN 값 제외)
df['age'].fillna(mean_age, inplace=True)

# age열의 첫 10개 데이터 출력(5행에 NaN값이 평균으로 대체)
print(df['age'].head(10))
```

```python
# 5-4 예제: 가장 많이 나타나는 값으로 바꾸기

import seaborn as sns

df = sns.load_dataset('titanic')

# embark_town 열의 829행의 NaN 데이터 출력
print(df['embark_town'][825:830])
print("\n")

# embark_town 열의 NaN 값을 승선도시 중에서 가장 많이 출현한 값으로 치환하기
most_freq = df['embark_town'].value_counts(dropna=True).idxmax()   # 가장 큰 값을 갖는 도시
print(most_freq)
print("\n")

df['embark_town'].fillna(most_freq, inplace=True)

# embark_town 열 829행의 NaN 데이터 출력(NaN 값이 most_freq 값으로 대체)
print(df['embark_town'][825:830])

```

- ex) df.replace("?", np.nan, inplace = True)

```python
# 예제 5-5: 이웃하고 있는 값으로 바꾸기 - 데이터셋의 특성상 이웃하는 데이터끼리 유사성을 갖는 경우

import seaborn as sns

df = sns.load_dataset('titanic')

# embark_town 열의 829행의 NaN 데이터 출력
print(df['embark_town'][825:830])
print("\n")

# embark_town 열의 NaN 값을 바로 앞에 있는 828행의 값으로 변경하기
df['embark_town'].fillna(method = 'ffill', inplace = True)
print(df['embark_town'][825:830])
```

## 5.2. 중복 데이터 처리

- 하나의 데이터셋에서 동일한 관측값이 2개이상 중복되는 경우 -> 중복데이터 삭제
- duplicated() 메소드 이용

```python
# 예제 5-6: 중복 데이터 확인

import pandas as pd

df = pd.DataFrame({'c1': ['a','a','b','a','b'],
                   "c2":[1,1,1,2,2],
                   "c3": [1,1,2,2,2]})

print(df)
print("\n")

# 데이터프레임 전체 행 데이터 중에서 중복값 찾기
df_dup = df.duplicated()
print(df_dup)
print("\n")

print("\n")
print("----------------------------------")
print("\n")

# 데이터프레임의 특정 열 데이터에서 중복값 찾기
col_dup = df['c2'].duplicated()
print(col_dup)
```

## 중복 데이터 제거

- drop_duplicates()
- 중복되는 행을 제거하고 고유한 관측값을 가진 행들만 남김
- subset 옵션에 열이름의 리스트 전달 가능 -> 데이터의 중복 여부 판별할 때 subset 옵션에 해당하는 열을 기준으로 판단

```python
# 예제 5-7: 중복 데이터 제거

import pandas as pd

df = pd.DataFrame({'c1': ['a','a','b','a','b'],
                   "c2":[1,1,1,2,2],
                   "c3": [1,1,2,2,2]})

print(df)
print("\n")

df2 = df.drop_duplicates()
print(df2)
print("\n")

# c2, c3열을 기준으로 중복행 제거
df3 = df.drop_duplicates(subset=['c2','c3'])
print(df3)
```

## 4-2 더미 변수

- 범주형데이터를 머신러닝 알고리즘에 바로 사용할 수 없는 경우 -> 컴퓨터가 인식 가능한 입력값으로 변환해야 한다
- 숫자 0 또는 1로 표현되는 더미변수를 사용
- 0 또는 1은 어떤 특성의 존재여부만 표시
- 원핫인코딩
- get_dummies(): 범주형 변수의 모든 고유값을 각각 새로운 더미변수로 변환



```python
# 예제 5-11: 더미 변수
# 예제 5-12: 원핫인코딩

from google.colab import drive
import pandas as pd

# 1. 구글 드라이브 마운트
drive.mount('/content/drive')

# 2. CSV 파일 경로
file_path = "/content/drive/MyDrive/[980-8]예제파일_파이썬 머신러닝 판다스 데이터 분석(개정판)/pandas-data-analysis-main/part5/data/auto-mpg.csv"

# 3. 데이터 불러오기
df = pd.read_csv(file_path, header=None)

# 열 이름 지정
df.columns = ['mpg', 'cylinders', 'displacement', 'horsepower', 'weight','acceleration',
              'model year', 'origin','name']

# horsepower 열의 누락데이터 삭제하고 실수형으로 변환
df['horsepower'].replace('?', np.nan, inplace=True)
df.dropna(subset=['horsepower'], axis=0, inplace=True)
df['horsepower'] = df['horsepower'].astype('float')

# np.histogram 함수로 3개의 bin으로 나누는 경계값의 리스트 구하기
count, bin_dividers = np.histogram(df['horsepower'], bins=3)
print(bin_dividers)

print("\n")
print("--------------------------------------------")
print("\n")

# 3개의 bin에 이름 지정
bin_names = ['저출력', '보통출력','고출력']

# pd.cut 함수로 각 데이터를 3개의 bin에 할당
df['hp_bin'] = pd.cut(x=df['horsepower'],
                      bins=bin_dividers,
                      labels=bin_names,
                      include_lowest=True)

# hp_bin 열의 범주형 데이터를 더미변수로 변환
horsepower_dummies = pd.get_dummies(df['hp_bin'], dtype = int)
print(horsepower_dummies.head(15))


print("\n")
print("--------------------------------------------")
print("\n")


# sklern 라이브러리 불러오기
from sklearn import preprocessing

# 전처리를 위한 encoder 객체 만들기
label_encoder = preprocessing.LabelEncoder()
onehot_encoder = preprocessing.OneHotEncoder()

# label encoder로 문자열 범주를 숫자형 범주로 변환
onehot_labeled = label_encoder.fit_transform(df['hp_bin'].head(15))
print(onehot_labeled)
print(type(onehot_labeled))

print("\n")
print("--------------------------------------------")
print("\n")

# 2차원 행렬로 형태 변경
onehot_reshaped = onehot_labeled.reshape(len(onehot_labeled), 1)
print(onehot_reshaped)
print(type(onehot_reshaped))

print("\n")
print("--------------------------------------------")
print("\n")

# 최소행렬로 변환
onehot_fitted = onehot_encoder.fit_transform(onehot_reshaped)
print(onehot_fitted)
print(type(onehot_fitted))
```

- Timestamp를 Period로 변환
- to_period()

```python
import pandas as pd

# 날짜 형식의 문자열로 구성되는 리스트 정의
dates= ['2019-01-01', '2020-03-01', '2021-06-01']

# 문자열의 배열을 판다스 timestamp로 변환
ts_dates = pd.to_datetime(dates)
print(ts_dates)
print("\n")

# Timestamp를 period로 변환
pr_day = ts_dates.to_period(freq='D')
print(pr_day)
pr_month = ts_dates.to_period(freq='M')
print(pr_month)
pr_year = ts_dates.to_period(freq='A')
print(pr_year)
```

## 6-2 시계열 데이터 만들기

- date_range(): 여러 개의 날짜가 들어있는 배열형태의 시계열 데이터를 만들 수 있음
- period_range(): 여러개의 기간이 들어있는 시계열 데이터 생성

```python
# 예제 5-17: TImestamp 배열 만들기

import pandas as pd

# Timestamp의 배열 만들기 - 월간격, 월의 시작일 기준
ts_ms = pd.date_range(start ='2019-01-01',
                      end=None,
                      periods = 6,
                      freq = 'MS',
                      tz = 'Asia/Seoul')
print(ts_ms)
print('\n')
print("--------------------------------------")
print("\n")

ts_me = pd.date_range('2019-01-01', periods=6,
                      freq='M',
                      tz='Asia/Seoul')
print(ts_me)

print('\n')
print("--------------------------------------")
print("\n")

# 분기(3개월) 간격, 월 마지막 기준
ts_3m = pd.date_range('2019-01-01', periods = 6,
                      freq = '3M',
                      tz = 'Asia/Seoul')
print(ts_3m)
```

```python
# 5-18: Period 배열 만들기

import pandas as pd

# Period 배열 만들기 - 1개월 길이
pr_m = pd.period_range(start='2019-01-01',
                       end = None,
                       periods = 3,
                       freq = 'M')

print(pr_m)

# Period 배열 만들기 - 1시간 길이
pr_h = pd.period_range(start = '2019-01-01',
                       end = None,
                       periods=3,
                       freq = 'H')
print(pr_h)
print("\n")

# period 배열 만들기 - 2시간 길이
pr_2h = pd.period_range(start = '2019-01-01',
                        end = None,
                        periods = 3,
                        freq = '2H')
print(pr_2h)
```

## 6-3 시계열 데이터 활용

- 연-월-일 날짜 데이터에서 일부를 분리하여 추출할 수 있음
- 날짜 인덱스 활용: 시계열 데이터에 대한 인덱싱과 슬라이싱이 편함

```python
# 예제 5-19: 날짜 데이터 분리

from google.colab import drive
import pandas as pd

# 1. 구글 드라이브 마운트
drive.mount('/content/drive')

# 2. CSV 파일 경로
file_path = "/content/drive/MyDrive/[980-8]예제파일_파이썬 머신러닝 판다스 데이터 분석(개정판)/pandas-data-analysis-main/part5/data/stock-data.csv"


# 3. 데이터 불러오기
df = pd.read_csv(file_path, header=0)

# 문자열인 날짜 데이터를 판다스 Timestamp로 변환
df['new_Date'] = pd.to_datetime(df['Date'])
print(df.head())
print("\n")

# dt 속성을 이용하여 new_Date 열의 연-월-일 정보를 년, 월, 일로 구분
df['Year'] = df['new_Date'].dt.year
df['Month'] = df['new_Date'].dt.month
df['Day'] = df['new_Date'].dt.day
print(df.head())

# Timestamp를 period로 변환하여 연-월-일 표기 변경하기
df['Date_yr'] = df['new_Date'].dt.to_period(freq='A')
df['Date_m'] = df['new_Date'].dt.to_period(freq='M')
print(df.head())

# 원하는 열을 새로운 행 인덱스로 지정
df.set_index('Date_m', inplace = True)
print(df.head())
```

```python
# 예제 5-20: 날짜 인덱스 활용

from google.colab import drive
import pandas as pd

# 1. 구글 드라이브 마운트
drive.mount('/content/drive')

# 2. CSV 파일 경로
file_path = "/content/drive/MyDrive/[980-8]예제파일_파이썬 머신러닝 판다스 데이터 분석(개정판)/pandas-data-analysis-main/part5/data/stock-data.csv"


# 3. 데이터 불러오기
df = pd.read_csv(file_path, header=0)

# 문자열인 날짜 데이터를 판다스 Timestamp로 변환
df['new_Date'] = pd.to_datetime(df['Date'])
df.set_index('new_Date', inplace= True)

print(df.head())
print("\n")
print(df.index)

# 날짜 인덱스를 이용하여 데이터 선택하기
df_y = df.loc['2018']
print(df_y.head())
print("\n")

df_ym = df.loc['2018-07']
print(df_ym)
print("\n")

df_ym_cols = df.loc['2018-07', 'Start':'High']
print(df_ym_cols)
print("\n")

df_ymd = df.loc['2018-07-02']
print(df_ymd)
print("\n")

df = df.sort_index()

df_ymd_range = df.loc['2018-06-20':'2018-06-25']
print(df_ymd_range)

# 시간 간격 계산. 최근 180-189일 사이의 값들만 선택하기
today = pd.to_datetime('2018-12-25')

df['time_delta'] = today - df.index

# 시간 간격을 인덱스로 설정
df.set_index('time_delta', inplace=True)

# 시간 간격 기준 오름차순 정렬
df.sort_index(inplace=True)

# 180일~189일 데이터 선택
df_180 = df.loc['180 days':'189 days']

print(df_180)
```
