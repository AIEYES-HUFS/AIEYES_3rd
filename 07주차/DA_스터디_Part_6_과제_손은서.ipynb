## 1. 함수 매핑

- lambda 함수: 이름 없이 한 줄로 정의하는 함수

### 1-1 개별 원소에 함수 매핑
#### 1-1-1 시리즈 원소에 함수 매핑

- apply(): 인자로 전달하는 매핑 함수에 시리즈의 모든 원소를 하나씩 입력하고 리턴값을 돌려받는다.
    - `Series객체.apply(매핑함수)`

```python
# -*- coding: utf-8 -*-

# 라이브러리 불러오기
import seaborn as sns

# titanic 데이터셋에서 age, fare 2개 열을 선택하여 데이터프레임 만들기
titanic = sns.load_dataset('titanic')
df = titanic.loc[:, ['age','fare']]
df['ten'] = 10
print(df.head())
```

```python
# 사용자 함수 정의
def add_10(n):                # 10을 더하는 함수
    return n + 10

def add_two_obj(a, b):        # 두 객체의 합
    return a + b

print(add_10(10))
print(add_two_obj(10, 10))
```

```python
# 시리즈 객체에 적용
sr1 = df['age'].apply(add_10)                  # n = df['age']의 모든
#df['age'](시리즈)의 모든 원소가 하나씩 add_10의 n으로 전달되고, 리턴값들이 시리즈 sr1에 저장됨
print(sr1.head())
print('\n')

# 시리즈 객체와 숫자에 적용: 2개의 인수(시리즈 + 숫자)
sr2 = df['age'].apply(add_two_obj, b=10)       # a=df['age']의 모든 원소, b=10
print(sr2.head())
print('\n')

# lambda 함수 활용: 시리즈 객체에 적용
sr3 = df['age'].apply(lambda x: add_10(x))     # x=df['age']
print(sr3.head())
```

#### 1-1-2 데이터프레임 원소에 함수 매핑

- `applymap()` :매핑 함수에 데이터프레임의 각 원소를 하나씩 넣어서 리턴값으로 돌려받음
    - 형식: `DataFrame객체.applymap(매핑함수)`
- 매핑?: 한 집합의 각 원소를 다른 집합의 원소에 일대일로 대응시키는 것
    - 매핑 함수를 적용한다??: 메소드가 객체의 원소를 매핑 함수의 인자로 전달하고, 함수가 반환한 값을 결과로 돌려받는 과정

```python
# -*- coding: utf-8 -*-

# 라이브러리 불러오기
import seaborn as sns

# titanic 데이터셋에서 age, fare 2개 열을 선택하여 데이터프레임 만들기
titanic = sns.load_dataset('titanic')
df = titanic.loc[:, ['age','fare']]
print(df.head())
print('\n')

# 사용자 함수 정의
def add_10(n):      # 10을 더하는 함수
    return n + 10

# 데이터프레임에 applymap()으로 add_10() 함수를 매핑 적용
#->df의 모든 원소를 add_10 함수에 하나씩 넣어서, 반환값으로 이루어진 새 데이터프레임을 만든다
df_map = df.applymap(add_10)
print(df_map.head())
```

### 1-2 시리즈 객체에 함수 매핑

#### 1-2-1 데이터프레임의 각 열에 함수 매핑

- `DataFrame객체.apply(매핑함수, axis=0)` : 모든 열을 하나씩 분리하여 매핑 함수의 인자로 각 열(시리즈)을 전달
    - 매핑 함수가 무엇을 반환하느냐에 따라 반환되는 객체의 종류가 다름
    - 시리즈를 입력받아 시리즈를 반환하는 함수: 각 열의 리턴값이 시리즈 형태로 반환, 이들이 하나의 데이터프레임으로 통합
    - 시리즈 →하나의 값 반환: 각 열의 리턴값이 하나의 값으로 반환되고 하나의 시리즈로 통합됨
        - 각 열의 이름: 시리즈의 인덱스
        - 함수가 반환하는 값: 각 인덱스에 매칭되는 데이터 값
- `Series객체.isnull()`: 시리즈 객체에서 누락 데이터를 찾고 그 결과를 불리언 시리즈 형태로 반환

```python
# -*- coding: utf-8 -*-

# 라이브러리 불러오기
import seaborn as sns

# titanic 데이터셋에서 age, fare 2개 열을 선택하여 데이터프레임 만들기
titanic = sns.load_dataset('titanic')
df = titanic.loc[:, ['age','fare']]
print(df.head())
print('\n')

# 사용자 함수 정의
def missing_value(series):          # 시리즈를 인자로 전달
    return series.isnull()          # 불린 시리즈를 반환

# 데이터프레임의 각 열을 인자로 전달하면 데이터프레임을 반환
result = df.apply(missing_value, axis=0)
print(result.head())
print('\n')
print(type(result))
```

```python
# -*- coding: utf-8 -*-

# 라이브러리 불러오기
import seaborn as sns

# titanic 데이터셋에서 age, fare 2개 열을 선택하여 데이터프레임 만들기
titanic = sns.load_dataset('titanic')
df = titanic.loc[:, ['age','fare']]
print(df.head())
print('\n')

# 사용자 함수 정의
def min_max(x):                  # 최대값 - 최소값
    return x.max() - x.min()

# 데이터프레임의 각 열을 인자로 전달하면 시리즈를 반환
result = df.apply(min_max)       # 기본값 axis=0
print(result)
print('\n')
print(type(result))
```

#### 1-2-2 데이터프레임의 각 행에 함수 매핑

- `DataFrame객체.apply(매핑함수, axis=1)` : 데이터프레임의 각 행을 매핑 함수의 인자로 전달

### 1-3 데이터프레임 객체의 함수 매핑

- `DataFrame객체.pipe(매핑함수)`: 데이터프레임 전체를 매핑 함수의 인자로 전달
    - 함수가 반환하는 리턴값에 따라 `pipe()`가 반환하는 객체의 종류가 결정
        - 데이터프레임을 반환하는 함수 →데이터프레임
        - 시리즈를 반환하는 함수 →
        - 개별 값을 반환하는 함수 →값
    - 객체 전체를 한 번에 전달

```python
# -*- coding: utf-8 -*-

# 라이브러리 불러오기
import seaborn as sns

# titanic 데이터셋에서 age, fare 2개 열을 선택하여 데이터프레임 만들기
titanic = sns.load_dataset('titanic')
df = titanic.loc[:, ['age','fare']]

# 각 열의 NaN 찾기 - 데이터프레임 전달하면 데이터프레임 반환
def missing_value(x):
    return x.isnull()

# 각 열의 NaN 개수 반환 - 데이터프레임 전달하면 시리즈 반환
def missing_count(x):
    return missing_value(x).sum()

# 데이터프레임의 총 NaN 개수 - 데이터프레임 전달하면 값 반환
def totoal_number_missing(x):
    return missing_count(x).sum()

# 데이터프레임에 pipe() 메소드로 함수 매핑
result_df = df.pipe(missing_value)
print(result_df.head())
print(type(result_df))
```

```python
result_series = df.pipe(missing_count)
print(result_series)
print(type(result_series))
```

```python
result_value = df.pipe(totoal_number_missing)
print(result_value)
print(type(result_value))
```

## 3. 필터링

> 필터링? 시리즈 또는 데이터프레임의 데이터 중에서 특정 조건식을 만족하는 원소만 따로 추출하는 개념
>

### 3-1 불린 인덱싱

> 조건식을 적용해서 각 원소의 참/거짓을 판별하고, 그 결과(불린 시리즈)로 참인 데이터만 선택하는 필터링 기법
>
- 형식: `DataFrame 객체[불린 시리즈]`
- 조건식: 비교 연산자를 사용한 식
    - &: AND 연산자
    - |: OR 연산자

```python
# -*- coding: utf-8 -*-

# 라이브러리 불러오기
import seaborn as sns

# titanic 데이터셋 로딩
titanic = sns.load_dataset('titanic')

# 나이가 10대(10~19세)인 승객만 따로 선택
mask1 = (titanic.age >= 10) & (titanic.age < 20)
# mask1이 참인 행만, 모든 열(:)을 선택해 df_teenage에 저장
df_teenage = titanic.loc[mask1, :]
print(df_teenage.head())
```

```python
# 나이가 10세 미만(0~9세)이고 여성인 승객만 따로 선택
mask2 = (titanic.age < 10) & (titanic.sex == 'female')
df_female_under10 = titanic.loc[mask2, :]
print(df_female_under10.head())
```

```python
# 나이가 10세 미만(0~9세) 또는 60세 이상인 승객의 age, sex, alone 열만 선택
mask3 = (titanic.age < 10) | (titanic.age >= 60)
df_under10_morethan60 = titanic.loc[mask3, ['age', 'sex', 'alone']]
print(df_under10_morethan60.head())
```

### 3-2 isin() 메소드 활용

- `isin()` : 특정 값을 가진 행들을 따로 추출할 수 있음
    - `isin()` 메소드에 데이터프레임의 열에서 추출하려는 값들로 만든 리스트를 전달
    - 형식: `DataFrame의 열 객체.isin(추출값의 리스트)`
    - 여러 개의 `==` 조건을 `|`로 이어 붙이는 것과 같은 결과를 내면서, 코드를 더 짧게 쓸 수 있음
- `pandas.set_option('옵션이름', 값)` : pandas 출력 설정을 바꾸는 함수

```python
# -*- coding: utf-8 -*-

# 라이브러리 불러오기
import seaborn as sns
import pandas as pd

# titanic 데이터셋 로딩
titanic = sns.load_dataset('titanic')

# IPython 디스플레이 설정 - 출력할 열의 개수
pd.set_option('display.max_columns', 10)    # 출력할 열의 개수 한도

# 함께 탑승한 형제 또는 배우자의 수가 3, 4, 5인 승객만 따로 추출 - 불린 인덱싱
#‘sibsp’ 열의 값이 3인지 판별한 불린 시리즈를 저장
mask3 = titanic['sibsp'] == 3
mask4 = titanic['sibsp'] == 4
mask5 = titanic['sibsp'] == 5
# 세 불린 시리즈를 |(OR)로 결합해서 하나라도 참인 행만 선택
df_boolean = titanic[mask3 | mask4 | mask5]
print(df_boolean.head())
```

```python
# isin() 메소드 활용하여 동일한 조건으로 추출
isin_filter = titanic['sibsp'].isin([3, 4, 5])
# isin_filter가 참인 행만 선택해서 df_isin에 저장
df_isin = titanic[isin_filter]
print(df_isin.head())
```

## 4. 데이터프레임의 합치기

### 4-1 데이터프레임 연결

- `pandas.concat(데이터프레임 리스트)` : 데이터프레임을 원소로 갖는 리스트를 전달하면 여러 개의 객체로 서로 연결해서 하나로 반환
    - 기존 객체의 형태를 유지하면서 이어 붙임
    - axis 옵션: 연결하는 방향
        - `axis=0` : 기본값, 행 방향으로 연결
        - `axis=1`: 열 방향으로 연결
    - join 옵션: 축의 라벨을 어떻게 맞출지 정함
        - `join=’outer’`: 기본값, 두 객체 라벨의 합집합을 기준으로…, 한쪽에만 이는 위치는 NaN이 됨
        - `join=’inner’` : 두 객체에 공통으로 있는 라벨의 교집합만 남김
    - `ignore_index=True`: 기존 행 인덱스를 무시하고 0부터 정수형 위치 인덱스를 새로 설정

```python
# -*- coding: utf-8 -*-

# 라이브러리 불러오기
import pandas as pd

# 데이터프레임 만들기
df1 = pd.DataFrame({'a': ['a0', 'a1', 'a2', 'a3'],
                    'b': ['b0', 'b1', 'b2', 'b3'],
                    'c': ['c0', 'c1', 'c2', 'c3']},
                    index=[0, 1, 2, 3])

df2 = pd.DataFrame({'a': ['a2', 'a3', 'a4', 'a5'],
                    'b': ['b2', 'b3', 'b4', 'b5'],
                    'c': ['c2', 'c3', 'c4', 'c5'],
                    'd': ['d2', 'd3', 'd4', 'd5']},
                    index=[2, 3, 4, 5])

print(df1, '\n')
print(df2, '\n')

# 2개의 데이터프레임을 위 아래 행 방향으로 이어 붙이듯 연결하기
result1 = pd.concat([df1, df2])
print(result1, '\n')
```

```python
# ignore_index=True 옵션 설정하기
result2 = pd.concat([df1, df2], ignore_index=True)
print(result2, '\n')
```

```python
# 2개의 데이터프레임을 좌우 열 방향으로 이어 붙이듯 연결하기
result3 = pd.concat([df1, df2], axis=1)
print(result3, '\n')
```

```python
# join='inner' 옵션 적용하기(교집합)
result3_in = pd.concat([df1, df2], axis=1, join='inner')
print(result3_in, '\n')
```

```python
# 시리즈 만들기
sr1 = pd.Series(['e0', 'e1', 'e2', 'e3'], name='e')
sr2 = pd.Series(['f0', 'f1', 'f2'], name='f', index=[3, 4, 5])
sr3 = pd.Series(['g0', 'g1', 'g2', 'g3'], name='g')

# df1과 sr1을 좌우 열 방향으로 연결하기
result4 = pd.concat([df1, sr1], axis=1)
print(result4, '\n')

# df2와 sr2를 좌우 열 방향으로 연결하기
result5 = pd.concat([df2, sr2], axis=1, sort=True)
print(result5, '\n')
```

```python
# sr1과 sr3을 좌우 열 방향으로 연결하기
result6 = pd.concat([sr1, sr3], axis=1)
print(result6, '\n')

result7 = pd.concat([sr1, sr3], axis=0)
print(result7, '\n')
```

### 4-2 데이터프레임 병합

- `pandas.merge(df_left, df_right, how='inner', on=None)` : 어떤 기준에 의해 두 데이터프레임을 병합
    - 키(key): 기준이 되는 열이나 인덱스
    - 키가 되는 열이나 인덱스는 양쪽 데이터프레임 모두 존재해야 함
    - `on` 옵션: 키로 사용할 열 이름을 지정
        - `on=None` (기본값): 두 데이터프레임에 공통으로 속하는 모든 열을 키로 병합
        - `on='열이름'`: 지정한 열을 키로 병합
    - `how` 옵션: 어떤 키 값의 행을 남길지 정함
        - `how='inner'` (기본값): 키 열의 데이터가 양쪽에 공통으로 존재하는 교집합만 추출
        - `how='outer'`: 키 값이 어느 한쪽에만 있어도 포함하는 합집합, 데이터가 없는 쪽의 열은 `NaN`이 됨
        - `how='left'`: 왼쪽 데이터프레임의 키 열에 속하는 데이터 값을 기준으로 병합
        - `how='right'`: 오른쪽 데이터프레임의 키 열에 속하는 데이터 값을 기준으로 병합

## 5. 그룹 연산

- 1단계 분할(split): 데이터를 특정 조건에 의해 분할
    - 판다스의 `groupby()` 메소드를 사용
- 2단계 적용(apply): 데이터를 집계, 변환, 필터링하는 데 필요한 메소드를 적용
- 3단계 결합(combine): 2단계의 처리 결과를 하나로 결합

### 5-1 그룹 객체 만들기(분할 단계)

#### 5-1-1 1개 열을 기준으로 그룹화

- `DataFrame객체.groupby(기준이 되는 열)` : 데이터프레임을 특정 열의 값을 기준으로 분할하여 그룹 객체를 반환
    - 기준 열의 값이 같은 행끼리 하나의 그룹이 됨
#### 5-1-2 여려 열을 기준으로 그룹화

- `DataFrame객체.groupby(기준이 되는 열의 리스트)` : 여러 열을 리스트로 전달하면 각 열이 가진 값으로 만들 수 있는 모든 조합을 키로 만들어 그룹을 나눔
    - 조합된 키의 형태는 튜플!
- `그룹객체.get_group((키1, 키2))`: 여러 열 기준의 그룹 객체에서는 키를 튜플로 입력해서 그룹 하나를 추출
- `len(객체)`: 객체의 원소 개수를 반환하는 파이썬 내장 함수
    - 데이터프레임에 쓰면 행의 개수를 반환
- `그룹객체.mean()`: 그룹별로 각 열의 평균값을 계산
- `그룹객체.get_group(키값)`: 그룹 객체에서 키 값에 해당하는 그룹 하나를 데이터프레임으로 추출

```python
# -*- coding: utf-8 -*-

# 라이브러리 불러오기
import pandas as pd
import seaborn as sns

# titanic 데이터셋에서 age, sex 등 5개 열을 선택하여 데이터프레임 만들기
titanic = sns.load_dataset('titanic')
df = titanic.loc[:, ['age','sex', 'class', 'fare', 'survived']]

#df의 행 개수를 출
print('승객 수:', len(df))
print(df.head())
print('\n')

# class 열을 기준으로 분할
grouped = df.groupby(['class'])
print(grouped)
```

```python
# 그룹 객체를 iteration으로 출력: head() 메소드로 첫 5행만을 출력
for key, group in grouped:
    print('* key :', key)
    print('* number :', len(group))
    print(group.head())
    print('\n')
```

```python
# class 열, sex 열을 기준으로 분할
grouped_two = df.groupby(['class', 'sex'])

# grouped_two 그룹 객체에 연산 메소드 적용
average_two = grouped_two.mean()
print(average_two)
print('\n')
print(type(average_two))
```

```python
# 개별 그룹 선택하기
group3 = grouped.get_group('Third')
print(group3.head())
```

### 5-2 그룹 연산 메소드(적용-결합 단계)

#### 5-2-1 데이터 집계

> 집계(aggregation)?? 그룹 객체에 연산을 적용해서 그룹별로 하나의 값으로 요약하는 과정
>
- 집계 기능을 내장한 기본함수: `mean()`, `max()`, `min()`, `sum()`, `count()`, `size()`, `var()`, `std()`, `describe()`, `info()`, `first()`, `last()`
    - `그룹객체.std()` : 그룹별로 각 열의 표준편차를 집계해서 데이터프레임으로 반환
- `그룹객체.agg(매핑함수)` : 사용자가 정의한 함수를 그룹 객체에 적용해서 그룹별로 집계
    - 집계 함수는 각 그룹의 각 열을 인자로 받아 하나의 값을 반환해야 함
- `그룹객체.agg([함수1, 함수2, 함수3, ...])`: 모든 열에 여러 함수를 일괄 적용
- `그룹객체.agg({'열1': 함수1, '열2': 함수2, ...})`: 열마다 다른 함수를 적용

```python
# -*- coding: utf-8 -*-

# 라이브러리 불러오기
import pandas as pd
import seaborn as sns

# titanic 데이터셋에서 age, sex 등 5개 열을 선택하여 데이터프레임 만들기
titanic = sns.load_dataset('titanic')
df = titanic.loc[:, ['age','sex', 'class', 'fare', 'survived']]

# class 열을 기준으로 분할
grouped = df.groupby(['class'], observed=False)

# 각 그룹에 대한 모든 열의 표준편차를 집계하여 데이터프레임으로 반환
std_all = grouped.std(numeric_only=True)
print(std_all)
print('\n')
print(type(std_all))
print('\n')

# 각 그룹에 대한 fare 열의 표준편차를 집계하여 시리즈로 반환
std_fare = grouped.fare.std()
print(std_fare)
print('\n')
print(type(std_fare))
```

```python
# 그룹 객체에 agg() 메소드 적용 - 사용자 정의 함수를 인자로 전달
def min_max(x):    # 최대값 - 최소값
    return x.max() - x.min()

# 수치형 열만 선택하여 최대값과 최소값의 차이를 계산
agg_minmax = grouped[['age', 'fare', 'survived']].agg(min_max)
print(agg_minmax.head())
```

```python
# 여러 함수를 각 열에 동일하게 적용하여 집계
agg_all = grouped.agg(['min', 'max'])
print(agg_all.head())
print('\n')


# 각 열마다 다른 함수를 적용하여 집계
agg_sep = grouped.agg({'fare':['min', 'max'], 'age':'mean'})
print(agg_sep.head())
```

#### 5-2-2 그룹 연산 데이터 변환

- `그룹객체.transform(매핑함수)` : 그룹별로 구분해서 함수를 적용하지만 그룹별 집계 대신 각 원소의 본래 행 인덱스와 열 이름을 기준으로 연산 결과를 반환
    
    →그룹 연산의 결과를 원본 데이터프레임과 같은 형태로 변형해서 정리
    
    - `agg()`와의 차이: `agg()`는 그룹별로 집계해서 그룹 수만큼의 행을 반환하고, `transform()`은 원본과 같은 개수의 행을 원래 순서로 반환

```python
# 그룹별 age 열의 평균 집계 연산
age_mean = grouped.age.mean()
print(age_mean)
print('\n')

# 그룹별 age 열의 표준편차 집계 연산
age_std = grouped.age.std()
print(age_std)
print('\n')

# 그룹 객체의 age 열을 iteration으로 z-score를 계산하여 출력
for key, group in grouped.age:
    group_zscore = (group - age_mean.loc[key])/age_std.loc[key]
    print('* origin :', key)
    print(group_zscore.head(3))    # 각 그룹의 첫 3개의 행 출력
    print('\n')
```

```python
# z-score를 계산하는 사용자 함수 정의
def z_score(x):
    return (x - x.mean())/x.std()

# transform() 메소드를 이용하여 age 열의 데이터를 z-score로 변환
age_zscore = grouped.age.transform(z_score)
print(age_zscore.loc[[1, 9, 0]])    # 1, 2, 3 그룹의 첫 데이터 확인(변환 결과)
print('\n')
print(len(age_zscore))              # transform 메소드 반환 값의 길이
print('\n')
print(age_zscore.loc[0:9])          # transform 메소드 반환 값 출력(첫 10개)
print('\n')
print(type(age_zscore))             # transform 메소드 반환 객체의 자료형
```

#### 5-2-3 그룹 객체 필터링

- `그룹객체.filter(조건식 함수)`: 그룹 객체에 조건식을 가진 함수를 전달하면, 함수의 반환값이 참인 그룹만 남김

```python
# 데이터 개수가 200개 이상인 그룹만을 필터링하여 데이터프레임으로 반환
grouped_filter = grouped.filter(lambda x: len(x) >= 200)
print(grouped_filter.head())
print('\n')
print(type(grouped_filter))
```

```python
# age 열의 평균이 30보다 작은 그룹만을 필터링하여 데이터프레임으로 반환
age_filter = grouped.filter(lambda x: x.age.mean() < 30)
print(age_filter.tail())
print('\n')
print(type(age_filter))
```

#### 5-2-4 그룹 객체에 함수 매핑

- `그룹객체.apply(매핑함수)` : 판다스 객체의 개별 원소를 특정 함수에 일대일로 매핑

```python
# 집계: 각 그룹별 요약 통계정보 집계
agg_grouped = grouped.apply(lambda x: x.describe())
print(agg_grouped)
```

```python
# z-score를 계산하는 사용자 함수 정의
def z_score(x):
    return (x - x.mean())/x.std()

age_zscore = grouped.age.apply(z_score)    # 기본값 axis=0
print(age_zscore.head())
```

```python
# 필터링 : age 열의 데이터 평균이 30보다 작은 그룹만을 필터링하여 출력
age_filter = grouped.apply(lambda x: x.age.mean() < 30)
print(age_filter)
print('\n')
for x in age_filter.index:
    if age_filter[x]==True:
        age_filter_df = grouped.get_group(x)
        print(age_filter_df.head())
        print('\n')
```

## 6. 멀티 인덱스

> 멀티 인덱스?? 행(열) 인덱스를 여러 레벨로 구성한 인덱스
>
- `DataFrame객체.loc[첫째레벨값]`: 멀티 인덱스에서 첫 번째 레벨의 값 하나만 지정해서 그 값을 가진 행들을 선택
    - 지정한 라벨: 인덱스로 나타지 않음
    - 나머지 라벨: 인덱스로 남음
- `DataFrame객체.loc[(첫째레벨값, 둘째레벨값)]`: 각 레벨에서 찾을 값을 튜플로 전달해서 두 레벨을 모두 만족하는 행을 선택
    - 인덱스가 모두 지정되면 그 행의 시리즈가 반환됨
- `DataFrame객체.xs(값, level=레벨이름 또는 레벨번호)`: cross-section(단면) 을 선택하는 인덱서
    - `loc` 와 달리 첫 번째가 아닌 라벨도 지정 가능

```python
# class 열, sex 열을 기준으로 분할
grouped = df.groupby(['class', 'sex'])

# 그룹 객체에 연산 메소드 적용
gdf = grouped.mean()
print(gdf)
print('\n')
print(type(gdf))
```

```python
# class 값이 First인 행을 선택하여 출력
print(gdf.loc['First'])
```

```python
# class 값이 First이고, sex 값이 female인 행을 선택하여 출력
print(gdf.loc[('First', 'female')])
```

```python
# sex 값이 male인 행을 선택하여 출력
print(gdf.xs('male', level='sex'))
```

## 7. 피벗

- 구성 요소:
    - 행 인덱스
    - 열 인덱스
    - 데이터 값
    - 데이터 집계 함수
- `pandas.pivot_table(데이터프레임, index='열이름', columns='열이름', values='열이름', aggfunc='집계함수')`: 데이터프레임을 피벗테이블로 변환하는 함수
    - `index`: 행 인덱스로 들어갈 열, 이 열의 값이 행 라벨이 됨
    - `columns`: 열 인덱스로 들어갈 열, 이 열의 값이 열 라벨이 됨
    - `values`: 표 안의 데이터 값으로 사용할 열
    - `aggfunc`: 같은 칸에 해당하는 데이터가 여러 개일 때, 하나의 값으로 합치는 집계 함수
        - ex. `'mean'`, `'sum'`, `'max'`

```python
# -*- coding: utf-8 -*-

# 라이브러리 불러오기
import pandas as pd
import seaborn as sns

# IPython 디스플레이 설정 변경
pd.set_option('display.max_columns', 10)    # 출력할 최대 열의 개수
pd.set_option('display.max_colwidth', 20)   # 출력할 열의 너비

# titanic 데이터셋에서 age, sex 등 5개 열을 선택하여 데이터프레임 만들기
titanic = sns.load_dataset('titanic')
df = titanic.loc[:, ['age','sex', 'class', 'fare', 'survived']]
print(df.head())
print('\n')

# 행, 열, 값, 집계에 사용할 열을 1개씩 지정 - 평균 집계
pdf1 = pd.pivot_table(df,              # 피벗할 데이터프레임
                     index='class',    # 행 위치에 들어갈 열
                     columns='sex',    # 열 위치에 들어갈 열
                     values='age',     # 데이터로 사용할 열
                     aggfunc='mean')   # 데이터 집계 함수

print(pdf1.head())
```

```python
# 값에 적용하는 집계 함수 2개 이상 지정 가능 - 생존율, 생존자 수 집계
pdf2 = pd.pivot_table(df,                    # 피벗할 데이터프레임
                      index='class',         # 행 위치에 들어갈 열
                      columns='sex',         # 열 위치에 들어갈 열
                      values='survived',     # 데이터로 사용할 열
                      aggfunc=['mean', 'sum'])   # 데이터 집계 함수

print(pdf2.head())
```

```python
# 행, 열, 값에 사용할 열을 2개 이상 지정 가능 - 평균 나이, 최대 요금 집계
pdf3 = pd.pivot_table(df,                        # 피벗할 데이터프레임
                      index=['class', 'sex'],    # 행 위치에 들어갈 열
                      columns='survived',        # 열 위치에 들어갈 열
                      values=['age', 'fare'],    # 데이터로 사용할 열
                      aggfunc=['mean', 'max'])   # 데이터 집계 함수

# IPython Console 디스플레이 옵션 설정
pd.set_option('display.max_columns', 10)    # 출력할 열의 개수 한도
print(pdf3.head())
print('\n')

# 행, 열 구조 살펴보기
print(pdf3.index)
print(pdf3.columns)
```

```python
# xs 인덱서 사용 - 행 선택(default: axis=0)
print(pdf3.xs('First'))    # 행 인덱스가 First인 행을 선택
```

```python
print(pdf3.xs(('First', 'female')))    # 행 인덱스가 ('First', 'female')인 행을 선택
```

```python
print(pdf3.xs('male', level='sex'))    # 행 인덱스의 sex 레벨이 male인 행을 선택
```

```python
print(pdf3.xs(('Second', 'male'), level=[0, 'sex']))    # Second, male인 행을 선택
#행 인덱스 레벨 0(‘class’)에서 ‘Second’를 가져오고, ‘sex’ 레벨에서 ‘male’을 가져옴
```

```python
# xs 인덱서 사용 - 열 선택(axis=1 설정)
print(pdf3.xs('mean', axis=1))    # 열 인덱스가 mean인 데이터를 선택
```

```python
print(pdf3.xs(('mean', 'age'), axis=1))    # 열 인덱스가 ('mean', 'age')인 데이터 선택
```

```python
print(pdf3.xs(('max', 'fare', 0),
              level=[0, 1, 2], axis=1))    # max, fare, survived=0인 데이터 선택
```
