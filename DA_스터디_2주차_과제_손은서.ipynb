## 시리즈

**개념**

- 시리즈는 데이터가 순차적으로 나열된 1차원 배열이다
- 인덱스와 value가 일대일 대응 된다는 점에서 키(k)-키(v) 쌍으로 이루어진 파이썬 딕셔너리와 구조가 비슷하다
- 인덱스는 데이터 값의 위치를 나타내는 역할을 한다

**시리즈 만들기**

- 딕셔너리 →시리즈: `pandas.Series(딕셔너리)`
    - 딕셔너리의 키(k) →시리즈의 인덱스, 값(v) →데이커 값
- 리스트 →시리즈: `pd.Series(리스트)`
- 튜플 →시리즈: `pd.Series(튜플)`

```python
#pandas 불러오기
import pandas as pd
dict_data= {'a':1, 'b':2, 'c':3}
#딕셔너리 ->시리즈로 변환
sr= pd.Series(dict_data)
print(type(sr))
#type()함수로 변수 sr의 자료형을 확인해서 출력
print('\n')
#결과 사이에 빈 줄을 하나 넣어주는 역할
print(sr)
```

### 인덱스 구조

- 정수형 위치 인덱스 and 인덱스 이름(인덱스 라벨)
- `Series객체.index`: 인덱스 배열 선택
- `Series객체.values`: 데이터 값 배열만을 선택

```python
import pandas as pd
list_data= ['2019-01-02', 3.14, 'abc', 100, True]
sr= pd. Series(list_data)
print(sr)
#데이터 값 배열은 원래 데이터인 list_data의 리스트 원소 배열 순서와 갖게 입력된다
inx= sr.index
val= sr.values
print(inx)
print('\n')
print(val)
```

### 행/열 삭제 — drop()

- 행 삭제: `DataFrame객체.drop(행 인덱스 또는 배열, axis=0)`  # 또는 axis 생략
- 열 삭제: `DataFrame객체.drop(열 이름 또는 배열, axis=1)`
    - 여러 개 삭제 시 리스트로 전달: `df.drop(['우현','인아'], axis=0)`
    - drop()은 기본적으로 새 객체를 반환한다 (원본 불변)
        - 원본을 바꾸려면 `inplace=True`
    - 열 삭제 시 `axis=1`을 빠뜨리면 오류 발생

```python
import pandas as pd
exam_data={'수학': [90,80,70], '영어': [98, 89, 95], '음악': [85, 95, 100], '체육': [100, 90, 90]}
df=pd.DataFrame(exam_data, index=['서준', '우현', '인아'])
print(df)
print('\n')
df2= df[:]
df2.drop('우현', inplace=True)
print(df2)
print('\n')
df3=df[:]
df3.drop(['우현', '인아'],inplace=True)
print(df3)
```

### 행 인덱스 제배열 -**reindex()**

- `DataFrame객체.reindex(새로운 인덱스 배열)`
- 기존에 없던 인덱스를 추가하면 해당 행은 NaN으로 채워진다.
- NaN 대신 값을 채우려면 `fill_value` 옵션 사용:
    - `df.reindex(new_index, fill_value=0)`

```python
import pandas as pd
dict_data={'c0':[1,2,3], 'c1':[4,5,6], 'c2':[7,8,9], 'c3':[10,11,12], 'c4':[13,14,15]}
df=pd.DataFrame(dict_data, index=['r0', 'r1', 'r2'])
print(df)
print('\n')
new_index=['r0','r1','r2','r3','r4']

ndf=df.reindex(new_index)
#그냥 바로 인덱스를 넣어도 되나?
print(ndf)
print('\n')
new_index=['r0','r1','r2','r3','r4']
ndf2=df.reindex(new_index, fill_value=0)
print(ndf2)
```

# 산술연산

**공통 원리:** 판다스 객체의 산술연산은 내부적으로 3단계를 거친다.

1. 행/열 인덱스를 기준으로 정렬
2. 같은 위치의 원소끼리 일대일 대응
3. 대응되는 원소끼리 연산 (대응 짝이 없으면 **NaN**)

## 시리즈 연산

### 시리즈 vs 숫자

- 시리즈와 숫자 연산: Series객체 + 연산자(+,-,*,/) + 숫자
- 시리즈의 모든 원소에 동일하게 숫자 연산이 적용되고, 인덱스는 그대로 유지된다

```python
import pandas as pd
student1=pd.Series({'국어':100, '영어':80, '수학':90})
print(student1)
print('\n')
percentage=student1/200
print(percentage)
print('\n')
print(type(percentage))
```

### 시리즈 vs 시리즈

- 시리즈와 시리즈 사이의 사칙연산을 처리하는 방법
- `Series1 + 연산자(+,-,*,/) + Series2`
- 같은 인덱스끼리 계산되며, 순서가 달라도 판다스가 자동으로 정렬 후 계산한다
- 한쪽에만 존재하는 인덱스, 또는 한쪽 값이 NaN인 경우 → 연산 결과는 NaN

```python
from numpy import subtract
import pandas as pd
student1=pd.Series({'국어':100, '영어':80, '수학':90})
student2=pd.Series({'수학':80, '국어':90, '영어':80})
print(student1)
print('\n')
print(student2)
print('\n')
a=student1+student2
s=student1-student2
m=student1*student2
d=student1/student2
print(type(d))
print('\n')
r=pd.DataFrame([a,s,m,d], index=['덧셈','뺄셈','곱셈','나눗셈'])
print(r)
```

```python
import numpy as np
import pandas as pd
student1=pd.Series({'국어':np.nan, '영어':80, '수학':90})
student2=pd.Series({'수학':80, '국어':90, '영어':80})
print(student1)
print('\n')
print(student2)
print('\n')
a=student1+student2
s=student1-student2
m=student1*student2
d=student1/student2
print(type(d))
print('\n')
r=pd.DataFrame([a,s,m,d], index=['덧셈','뺄셈','곱셈','나눗셈'])
print(r)
```

### 연산 메소드

- 연산 메소드 사용(시리즈와 시리즈 덧셈): `Series1.add(Series2, fill_vaule=0)`
- NaN 대신 지정한 값으로 채운 뒤 연산 → 짝이 없는 경우도 계산 가능해진다
- 단, 0으로 나누면 결과는 `inf`(무한대)가 될 수 있음

```python
import numpy as np
import pandas as pd
student1=pd.Series({'국어':np.nan, '영어':80, '수학':90})
student2=pd.Series({'수학':80, '국어':90, '영어':80})
print(student1)
print('\n')
print(student2)
print('\n')
sa=student1.add(student2, fill_value=0)
sb=student1.sub(student2, fill_value=0)
sm=student1.mul(student2, fill_value=0)
sd=student1.div(student2, fill_value=0)

r=pd.DataFrame([sa,sb,sm,sd], index=['덧셈','뺄셈','곱셈','나눗셈'])
print(r)
```

## 데이터프레임 연산

### 데이터프레임 vs 숫자

- 데이터프레임과 숫자 연산: `DataFrame 객체 + 연산자(+,-,*,/) + 숫자`
- 모든 원소에 동일하게 적용
- 데이터프레임의 형태는 그대로 유지

```python
import pandas as pd
import seaborn as sns
titanic=sns.load_dataset('titanic')
df=titanic.loc[:,['age','fare']]
print(df.head())
print('\n')
print(type(df))
print('\n')
a=df+10
print(a.head())
print('\n')
print(type(a))
```

### 데이터프레임 vs 데이터프레임

- 데이터프레임의 연산자 활용: `DataFrame1 + 연산자(+,-,*,/) + DataFrame2`
- 같은 행·열 위치의 원소끼리 계산
- 어느 한쪽이라도 원소가 없거나 NaN이면 결과는 NaN

```python
import pandas as pd
import seaborn as sns
titanic=sns.load_dataset('titanic')
df=titanic.loc[:,['age','fare']]
print(df.tail())
print('\n')
print(type(df))
print('\n')
a=df+10
print(a.tail())
print('\n')
s=a-df
print(s.tail())
print('\n')
print(type(s))
```
